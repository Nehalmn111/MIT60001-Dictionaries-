```python
def find_grades(grades, students):
  """ grades is a dict maping student names(str) to grade
  students is a list of student names
  returns a list containing the grades for students"""
  Lnew = []
  for elem in students:
    grade = grades[elem]
    Lnew.append(grade)
  return Lnew
```

```python

def find_in_L(Ld, k):
  """Ld is the list of Dicts
    k is an int
    Returns True if k is a key in any dicts of Ld and False Otherwise"""
  for d in Ld:
    #d is k1,v1
    if k in d:
      return True
    return False
```

```python
song = ["sun", "is", "bright", "sun", "is", "warm", "sun"]

word_count = {}

for word in song:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1

print(word_count)

most_common = max(word_count, key=word_count.get)

print("Most common word:", most_common)
print("It appears:", word_count[most_common], "times")
```

```python
song = "RAH RAH AH AH AH ROM MAH RO MAH MAH"

def generate_word_dict(song):
    song_words = song.lower() #convert all the lower.
    words_list = song_words.split() #Split the spaces.
    word_dict = {}

    for w in words_list: #(can iterate over list of words)
        if w in word_dict:
            word_dict[w] += 1
        else:
            word_dict[w] = 1

    return word_dict
```

```python
word_dict = {'rah': 2, 'ah': 3, 'rom': 1, 'mah': 3, 'ro': 1}

def find_frequent_word(word_dict):
    words = []

    highest = max(word_dict.values()) # grab all of the values from the library.

    for k, v in word_dict.items():
        if v == highest:
            words.append(k)

    return (words, highest)
```
